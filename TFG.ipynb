```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Rango de valores de k a evaluar
k_values = range(2, 13)
distortions = []

# Calcular la distorsión para cada valor de k
for k in k_values:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X_scaled)
    distortions.append(model.inertia_)

# Elegir el valor óptimo de k (puedes cambiar este valor manualmente si lo ves en el gráfico)
k_optimo = 6
distorsion_optima = distortions[k_optimo - k_values.start]

# Graficar el método del codo
plt.figure(figsize=(8, 6))
plt.plot(k_values, distortions, marker='o', label='Distorsión')
plt.axvline(x=k_optimo, color='black', linestyle='--', label=f'k óptimo = {k_optimo}')
plt.xticks(k_values)
plt.xlabel('Número de clusters (k)')
plt.ylabel('Distorsión')
plt.title('Método del codo - Distorsión (K-Means)')
plt.legend()
plt.grid(True)
plt.savefig('elbow_method.svg', bbox_inches="tight")
plt.show()
```

```python
scores = []
ks = range(2, 11)

for k in ks:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    scores.append(score)

# Mostrar resultados
plt.plot(ks, scores, marker='o')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Coeficiente de silueta')
plt.title('Selección del número óptimo de clusters')
plt.grid(True)
plt.savefig('elbow_silhouette.svg', bbox_inches="tight")
plt.show()

# Mejor k
best_k = ks[scores.index(max(scores))]
print(f"Mejor número de clusters según Silhouette Score: {best_k}")
```

```python
import matplotlib.pyplot as plt

# Nombres de los algoritmos
algoritmos = ['K-Means', 'Affinity Propagation', 'Mean Shift', 'DBSCAN', 'HDBSCAN']

# Puntuaciones Davies-Bouldin (affinity hecho,
puntuaciones_db = [0.79, 0.84, 0.56, 1.45, 1.57]

# Crear el gráfico de barras
plt.figure(figsize=(8, 6))
barras = plt.bar(algoritmos, puntuaciones_db, color='skyblue', edgecolor='black')

# Añadir etiquetas a las barras
for barra in barras:
    altura = barra.get_height()
    plt.text(barra.get_x() + barra.get_width() / 2, altura + 0.02, f'{altura:.2f}',
             ha='center', va='bottom', fontsize=10)

# Añadir título y etiquetas
plt.title('Índice Davies-Bouldin por algoritmo de clustering', fontsize=14)
plt.ylabel('Índice Davies-Bouldin')
plt.ylim(0, max(puntuaciones_db) + 0.2)

# Mostrar gráfico
plt.tight_layout()
plt.grid(False)
plt.savefig('comparativa_davies_bouldin.svg', bbox_inches="tight")
plt.show()
```
